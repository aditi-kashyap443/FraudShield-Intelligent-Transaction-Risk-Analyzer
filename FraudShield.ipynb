```python
# Machine Learning

## Step 1: Separate Features and Target Variable

### Why?

#Machine learning models learn the relationship between input features (X) and the target variable (y).

#- Features (X): Information about each transaction.
#- Target (y): Fraud status that the model will predict.

#Separating X and y is the first step before splitting the dataset and training the model.
```

```python
# Features (Independent Variables)
X = df.drop("Is Fraud?", axis=1)

# Target (Dependent Variable)
y = df["Is Fraud?"]
```

```python
# Step 4: Understanding Class Imbalance

## Objective

#Class imbalance occurs when one class contains significantly more samples than the other.

#In fraud detection, genuine transactions greatly outnumber fraudulent transactions.

#This imbalance can cause machine learning models to become biased toward predicting the majority class.

#Therefore, it is important to identify and handle class imbalance before model training.
```

```python
print(y_train.value_counts())

print("\nPercentage:\n")

print((y_train.value_counts(normalize=True)*100).round(2))
```

```python
# Step 5: Logistic Regression

## Objective

#Logistic Regression is used as the baseline classification model.

#Although its name contains "Regression", it is mainly used for binary classification problems such as fraud detection.

#The model learns the relationship between the transaction features and the fraud label to predict whether a new transaction is fraudulent or genuine.

#Since the dataset is highly imbalanced, the parameter `class_weight="balanced"` is used so that fraudulent transactions receive more importance during training.
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
# Create the Logistic Regression model
lr_model = LogisticRegression(
    class_weight="balanced",
    random_state=42,
    max_iter=1000
)

# Train the model
lr_model.fit(X_train, y_train)
```

```python
# Step 6: Make Predictions

## Objective

#After training, the model is used to predict the fraud status of unseen transactions from the test dataset.

#These predictions will later be compared with the actual labels to evaluate the model's performance.
```

```python
# Predict on the test dataset
y_pred = lr_model.predict(X_test)
```

```python
# Step 7.1: Confusion Matrix

## Objective

#A Confusion Matrix summarizes the performance of a classification model by comparing the predicted labels with the actual labels.

#It shows:

#- True Positives (TP)
#- True Negatives (TN)
#- False Positives (FP)
#- False Negatives (FN)

#This helps us understand not only how many predictions were correct, but also the types of mistakes made by the model.
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay(confusion_matrix=cm).plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
# Step 7.2: Classification Report

## Objective

#The Classification Report summarizes the key evaluation metrics for a classification model:

#- Precision
#- Recall
#- F1-Score
#- Support

#These metrics provide a better evaluation than accuracy for imbalanced datasets like fraud detection.
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
# Observation

#- The Logistic Regression model achieved an overall accuracy of 88%.

#- For genuine transactions, the model performed well with high precision and recall.

#- For fraudulent transactions:
 # - Recall = 65%, meaning the model detected around two-thirds of fraud cases.
  #- Precision = 1%, indicating that many genuine transactions were incorrectly flagged as fraud.

#- This resulted in a large number of false positives.

#- Although Logistic Regression serves as a good baseline model, it is not sufficient for this highly imbalanced fraud detection problem.

#- More advanced models such as Random Forest or XGBoost are expected to improve performance.
```
